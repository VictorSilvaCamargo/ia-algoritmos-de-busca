```python
import numpy as np

# Classe node, é um vértice que possui um valor de utilidade e uma lista
# com a indicação de quem sãi os seus filhos

class Node:
  # método construtor
  def __init__(self, utility=None):
    self.utility = utility # valor de utilidade do vértice
    self.children = [] # lista de filhos ou sucessores


  def add_child(self, child):
    self.children.append(child)

  def is_terminal(self):
    if len(self.children) == 0:
      return True
    else:
      return False

  def get_children(self):
    return self.children

  def get_utility(self):
    return self.utility
```

```python
A = Node(0)
B = Node(10)
C = Node(5)

A.add_child(B)
A.add_child(C)

B.get_utility()
```

```python
# Algoritimo MINIMAX
# Recebe como entrada:
# - Um objeto do tipo "node"
# - Profundidade do Node na árvore (útil para saber se é max ou min)

def minimax(node, depth = 1):
  # quando  jogo começa (depth = 1), a primeira jogada é do Max
  if depth % 2 == 1: # profundidade ímpar
    agent = "MAX"
  else:
    agent = "MIN"

  if node.is_terminal():
    return node.get_utility()

  elif agent == "MAX":
    return max_value(node, depth)
  else:
    return min_value(node, depth)


# Busca recursivamente pelo maior valor  entre os sucessores
def max_value(node, depth):
  value = -np.inf
  for sucessor in node.get_children():
    value = max(value, minimax(sucessor, depth + 1))
  return value

def min_value(node, depth):
  value = np.inf
  for sucessor in node.get_children():
    value = min(value, minimax(sucessor, depth + 1 ))
  return value

```

```python
# Grafo 1
A = Node()
B = Node()
C = Node()
D = Node(2)
E = Node(7)
F = Node(1)
G = Node(8)

A.add_child(B)
A.add_child(C)

B.add_child(D)
B.add_child(E)

C.add_child(F)
C.add_child(G)

result = minimax(A)

print(f"O melhor resultado possível para MAX será {result}")
```
