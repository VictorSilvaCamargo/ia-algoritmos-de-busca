```python
#Busca em Parametro (DFS)
# Parametro de entrada
# graph: grafo representado por uma lista de adjacencias
# start: vétices de início da busca
# start: vértice de destino na busca

def DFS(graph, start, goal):
  # lista com os vértices já visitados
  visited = []

  # pilha com os vértices na fronteira e que são candidatos à exploração
  fringe = [start]

  # pilha que armazena os caminhos dos vértices candidatos até a raiz
  path = [[start]]

  # Entra em um laço de repetição equanto existirem elementos na pilha
  # (candidatos à expanção)
  while fringe:
    # Desempilha o item no topo e o associa para o vértice em visualisções
    current_node = fringe.pop()

    # Caminho da raiz até o current_node
    current_path = path.pop()

    if current_node not in visited:
      # Adiciona o vértice atual na lista de vértices visitados
      visited.append(current_node)

      # Verifica se o vértice atual é o objetivo
      if current_node == goal:
        return f'O Caminho encontrado foi {current_path}' # retorna o caminho da raiz até o vértice

      # Caso não seja o objetivo, é preciso adicionar seus filhos(adjacentes)
      # N a pilha de candidatos
      # É importante ordenar os filhos para garantir o início da visita mais a esquerda
      for adj_node in sorted(graph[current_node], reverse=True):

        # Adiciona o vértice adjacente na fronteira
        fringe.append(adj_node)

        # Adiciona o caminho até o vértice adjacente
        path.append(current_path + [adj_node])
  return None

```

```python
#Busca em Parametro (BFS)
# Parametro de entrada
# graph: grafo representado por uma lista de adjacencias
# start: vétices de início da busca
# start: vértice de destino na busca

def BFS(graph, start, goal):
  # lista com os vértices já visitados
  visited = []

  # fila com os vértices na fronteira e que são candidatos à exploração
  fringe = [start]

  # fila que armazena os caminhos dos vértices candidatos até a raiz
  path = [[start]]

  # Entra em um laço de repetição equanto existirem elementos na fila
  # (candidatos à expanção)
  while fringe:
    # tira o item do inicio da fila
    current_node = fringe.pop(0)

    # Caminho da raiz até o current_node
    current_path = path.pop(0)

    if current_node not in visited:
      # Adiciona o vértice atual na lista de vértices visitados
      visited.append(current_node)

      # Verifica se o vértice atual é o objetivo
      if current_node == goal:
        return f'O Caminho encontrado foi {current_path}' # retorna o caminho da raiz até o vértice

      # Caso não seja o objetivo, é preciso adicionar seus filhos(adjacentes)
      # N a pilha de candidatos
      # É importante ordenar os filhos para garantir o início da visita mais a esquerda
      for adj_node in sorted(graph[current_node]):

        # Adiciona o vértice adjacente na fronteira
        fringe.append(adj_node)

        # Adiciona o caminho até o vértice adjacente
        path.append(current_path + [adj_node])
  return None

BFS(G2, 'S', 'G')
```

```python
# Busca em Parametro (UCS)
# Parametro de entrada
# graph: grafo representado por uma lista de adjacencias
# start: vétices de início da busca
# start: vértice de destino na busca

# importa uma estrutura de fila de prioridades
import heapq

def UCS(graph, start, goal):
  # lista com os vértices já visitados
  visited = []

  # fila com os vértices na fronteira e que são candidatos à exploração
  fringe = []
  # fringe = [(0, start)]
  heapq.heappush(fringe, (0, start))

  # fila de prioridades que armazena os caminhos dos vertices candidatos ate a raiz
  path = []
  heapq.heappush(path, (0,[start]))


  # Entra em um laço de repetição equanto existirem elementos na pilha
  # (candidatos à expanção)
  while fringe:
    # tira o item do inicio da fila
    item = heapq.heappop(fringe) # retorna uma tupla( cost, node)
    current_cost = item[0] # custo até chegar ao vertice
    current_node = item[1] # vertice em visitação

    # Caminho da raiz até o current_node
    current_path = heapq.heappop(path)[1]


    if current_node not in visited:
      # Adiciona o vértice atual na lista de vértices visitados
      visited.append(current_node)

      # Verifica se o vértice atual é o objetivo
      if current_node == goal:
        return f'O Caminho {current_path} e o custo é {current_cost}'

      # Caso não seja o objetivo, é preciso adicionar seus filhos(adjacentes)
      # N a pilha de candidatos
      # É importante ordenar os filhos para garantir o início da visita mais a esquerda
      for adj_node,adj_cost in sorted(graph[current_node].items()):
        if adj_node not in visited:
          # Adiciona o custo acumulado e o vértice adjacente na fila de prioridades
          heapq.heappush(fringe, (current_cost + adj_cost, adj_node))

          # Adiciona os caminhos completos até o adj_node e o custo
          heapq.heappush(path,(current_cost + adj_cost, current_path + [adj_node]))

          # Distancia de path e o fringe é ter o caminho completo
          # e não apenas os vértices

  return None

```

```python
def DFS_Limited(graph, start, goal, limit):
    # lista com os vértices já visitados
    visited = []

    # pilha com os vértices na fronteira e que são candidatos à exploração
    fringe = [(start, 0)]  # Adicionando a profundidade junto ao nó

    # pilha que armazena os caminhos dos vértices candidatos até a raiz
    path = [[start]]

    # Entra em um laço de repetição enquanto existirem elementos na pilha
    # (candidatos à expansão)
    while fringe:
        # Desempilha o item no topo e o associa para o vértice em visualizações
        current_node, depth = fringe.pop()  # Obtendo o nó e a profundidade

        # Caminho da raiz até o current_node
        current_path = path.pop()

        if current_node not in visited:
            # Adiciona o vértice atual na lista de vértices visitados
            visited.append(current_node)

            # Verifica se o vértice atual é o objetivo
            if current_node == goal:
                return f'O Caminho encontrado foi {current_path}'  # retorna o caminho da raiz até o vértice

            # Caso não seja o objetivo, é preciso adicionar seus filhos (adjacentes)
            # N a pilha de candidatos
            if depth < limit:  # Adiciona apenas se a profundidade for menor que o limite
                for adj_node in sorted(graph[current_node], reverse=True):
                    fringe.append((adj_node, depth + 1))
                    path.append(current_path + [adj_node])

    return None  # Retorna None se nenhum caminho foi encontrado
```
